# Data quality in ATLAS Open Data: comparing three run selections

Every collision ATLAS records is a measurement of the detector as much as of the physics. Calorimeter modules draw noise , subsystems drop out of the readout, events are occasionally written out incomplete and none of that is distinguishable from real physics until someone checks. The standard check is the **Good Runs List (GRL)**: a list of the luminosity blocks in which every
subsystem was verified to be working, with everything outside it discarded before analysis begins.

Discarding data is not free, and the threshold for "good enough" is a judgement rather than a measurement. This notebook takes the GRL as a reference and compares it against two alternative selections, labelled **IG** and **IF**, asking one question in as many variables as possible:

> **Where do the three selections disagree, and does the disagreement look like physics or like a
> broken detector?**

The method is deliberately simple. Distributions from all three selections are overlaid and normalised to unit area, so the trivial difference a filter keeps fewer events drops out and only a change in *shape* remains. A shape difference means the filter is removing a particular kind of event, which is exactly what it is there to do; the interesting part is finding out which kind.

## The two streams

Two data streams are used throughout, since they come from different triggers and probe different subsystems.

**The dilepton stream.** The Z peak at 91 GeV acts as a calibration standard: its position and width are known, so any shift between selections is a direct statement about lepton reconstruction quality. For same-sign pairs there is no peak to anchor to and the spectrum is smooth background, which makes any structure appearing in it immediately suspicious.

**The QCD jet stream.** Missing transverse energy and jet multiplicity are the sensitive quantities here. Physical events have MET of order tens of GeV; a noise burst or a truncated event can produce thousands, which is why several plots deliberately span ranges no real event should reach.

The skim names are set once, as `LEP_SKIM` and `JET_SKIM` in the sample-building cell, and the rest of the notebook refers to the two streams by role rather than by skim string. Everything downstream the sample keys, the selection logic, the plots is unaffected by what those two strings contain.

## Looking at the detector directly

Alongside the one-dimensional comparisons, $\eta-\phi$ maps show *where* on the detector objects were reconstructed. Faults are localised in hardware, so they appear as localised features: a blank patch where a module is dead, a bright spot where a cell is manufacturing objects out of noise, thin bands where the cracks and service gaps genuinely are. None of that survives projection onto a
single axis, where a hotspot is smeared into the bulk and vanishes.

An effect that persists across every configuration is a property of the data; one that appears in exactly one slice is more likely a statistical accident of that slice.

## Three data-quality selections

Here is the concept the whole notebook is built around.

Not every collision ATLAS records is usable. Sometimes a calorimeter module is drawing noise, a detector subsystem is out of the data acquizition, or an event is written out incompletely. Before anyone does physics with the data, it is filtered and the standard filter is the **Good Runs List (GRL)**,
an explicit list of luminosity blocks (roughly one-minute intervals) during which every subsystem was confirmed to be operating nominally. Anything outside the GRL is thrown away.

That filter is not free. It costs luminosity, and the choice of what counts as "good enough" is a judgement call. This notebook compares the GRL baseline against two alternative selections, labelled **IG** and **IF**, to see what the different choices actually do to the data.

The question we are asking, over and over in different variables, is:

> **Where do the three selections disagree, and does the disagreement look like physics or like a
> broken detector?**

If all three give identical normalised distributions, the extra filtering is removing events but not changing the *character* of what survives. If one of them shows a bump, a tail, or a hotspot the others don't, that excess is a direct picture of what the filter is there to catch.

Two skims are loaded for each selection `2bjets` for the QCD jet stream and `2to4lep` for the dilepton stream since the two object types come from different triggers and are compared separately.

```python
atom.set_release('2025e-13tev-beta')

# GRL data samples - the standard Good Runs List selection, our reference
data_samples_1 = atom.build_data_dataset('2bjets', name="qcdjets_GRL", protocol='https')
data_samples_2 = atom.build_data_dataset('2to4lep', name="2lep_GRL", protocol='https')

# IG data samples - first alternative selection
data_samples_3 = atom.build_data_dataset('2bjets', name="qcdjet_IG", protocol='https')
data_samples_4 = atom.build_data_dataset('2to4lep', name="2lep_IG", protocol='https')

# IF data samples - second alternative selection
data_samples_5 = atom.build_data_dataset('2bjets', name="qcdjets_IF", protocol='https')
data_samples_6 = atom.build_data_dataset('2to4lep', name="2lep_IF", protocol='https')

samples = {**data_samples_1, **data_samples_2,**data_samples_3, **data_samples_4,**data_samples_5, **data_samples_6}
```

## Overlaying the three selections

`p3` is the workhorse of the notebook: give it a variable name and a range, and it overlays GRL, IG and IF on the same axes so you can see immediately where they part company.

The important detail is that each histogram is **normalised to unit area** before plotting. This is deliberate. The three selections keep different numbers of events by construction that is the whole point of a filter so comparing raw counts would only tell us what we already know. Normalising
removes the trivial difference and leaves the interesting one: *does the shape change?*

A shape difference means the filter is removing a specific kind of event, not just fewer events. That is the signal we are hunting for.

The log y-axis matters here too. Data-quality problems live in the tails a handful of pathological events with absurd momentum or missing energy and on a linear scale those are invisible next to a bulk distribution millions of events tall.

```python
def p3(var, name, axis_name, low, high, NB, title,ww):
  bins = np.linspace(low, high, NB)
  bin_centers = (bins[:-1] + bins[1:]) / 2
  bin_width = bins[1] - bins[0]
  xerr = bin_width / 2      # marks bin extent, not an uncertainty

  # Accumulate one histogram across all chunks held in the list DF
  def get_hist(DF):
    Hr = np.zeros(len(bins) - 1)
    for i in range(len(DF)):
      y = DF[i][f'{var}']
      y_np = np.asarray(y)
      hi, _ = np.histogram(y_np, bins=bins)
      Hr += hi
      return Hr     # NOTE: inside the loop - see the notes below this cell

  # ww selects which of the two data streams to compare
  if ww:
    DF1 = lepdata_GRL; DF2 = lepdata_IG; DF3 = lepdata_IF
  else:
    DF1 = qcddata_GRL; DF2 = qcddata_IG; DF3 = qcddata_IF

  Hr1 = get_hist(DF1)
  Hr2 = get_hist(DF2)
  Hr3 = get_hist(DF3)
  # Normalise to unit area: the three selections keep different numbers of events by
  # construction, so only the SHAPE difference is informative.
  Hr1 = Hr1 / np.sum(Hr1)
  Hr2 = Hr2 / np.sum(Hr2)
  Hr3 = Hr3 / np.sum(Hr3)
  plt.errorbar(bin_centers, Hr1, xerr=xerr, fmt='.', color='black', markersize=4, capsize=2, label=f'GRL')
  plt.errorbar(bin_centers, Hr2, xerr=xerr, fmt='+', color='red', markersize=4, capsize=2, label=f'IG')
  plt.errorbar(bin_centers, Hr3, xerr=xerr, fmt='x', color='blue', markersize=4, capsize=2, label=f'IF')
  plt.xlabel(f'{axis_name}')
  plt.ylabel('Counts')
  plt.title(f'{title}')
  plt.yscale('log')   # data-quality problems live in the tails
  plt.legend()
  plt.show()
  return 0
```

<a id='mp'></a>

## $\eta - \phi$ maps: looking at the detector directly

If `p3` asks *what* is different, this asks *where*.

An $\eta - \phi$ map is a picture of the detector unrolled flat: pseudorapidity along one axis, azimuthal angle around the beam along the other. Every reconstructed object gets placed at the point it was seen. Because the detector is (by design) azimuthally symmetric, a healthy map is smooth and featureless in $\phi$, with only the expected structure in $\eta$.

```python
def mp(DF, var1, var2, title, low1, high1, NB1, low2, high2, NB2):
    bins1 = np.linspace(low1, high1, NB1)
    bins2 = np.linspace(low2, high2, NB2)
    H = np.zeros((len(bins1) - 1, len(bins2) - 1))

    # accumulate a 2D histogram of var1 vs var2 across all processed files
    for i in range(len(DF)):
        x = DF[i][f'{var1}']
        y = DF[i][f'{var2}']

        # flatten nested (per-object) arrays into flat arrays of values before histogramming
        x_np = np.asarray(ak.flatten(x, axis=None)) if hasattr(x, 'layout') else np.asarray(x)
        y_np = np.asarray(ak.flatten(y, axis=None)) if hasattr(y, 'layout') else np.asarray(y)

        h, _, _ = np.histogram2d(x_np, y_np, bins=[bins1, bins2])
        H += h

    plt.figure(figsize=(8, 6))

    # mask empty bins so they appear blank rather than as a solid colour
    # (a dead region and a quiet one must not look the same when dead regions are the target)
    H_masked = np.ma.masked_where(H == 0, H)
    # Log colour scale: a noisy cell can sit orders of magnitude above its neighbours
    mesh = plt.pcolormesh(bins1, bins2, H_masked.T, norm=LogNorm(vmin=1, vmax=H.max()), cmap='coolwarm')
    plt.colorbar(mesh, label='Counts')

    plt.xlabel(r'$\eta$')
    plt.ylabel(r'$\phi$')
    plt.title(f'{title}')
    plt.show()
    return 0
```

<a id='mass'></a>

## Invariant mass

The invariant mass of a system of particles is the one quantity every observer agrees on, whatever frame they are in. For a pair of leptons it comes straight from the relativistic energy-momentum
relation:

$$ m^2 = E^2 - |\vec{p}|^2 = \left(\sum_i E_i\right)^2 - \left|\sum_i \vec{p}_i\right|^2 $$

Note the order: sum the four-momenta *first*, then take the mass. Doing it the other way round gives you something else entirely.

The components are built from the detector's natural coordinates. Transverse momentum and azimuth give $p_x = p_T\cos\phi$ and $p_y = p_T\sin\phi$ directly. The longitudinal component uses the identity $\theta = 2\arctan(e^{-\eta})$ relating pseudorapidity to polar angle, so that $p_z = p_T/\tan\theta$
which is the same thing as the more familiar $p_T \sinh\eta$, just written in terms of what the ntuple stores.

The `ak.where(Mass_2 > 0, ...)` guard handles a real effect, not a hypothetical one: finite detector resolution can push a measured $E^2 - p^2$ slightly negative for a genuinely massless or near-massless system, and taking the square root would give `NaN`. Setting those to zero keeps them in the histogram
where you can see them, at the bottom of the range.

```python
def mass(E, lep_pt, eta, phi):
    # sum the 4-momentum components of the lepton pair across leptons in the event
    # (four-momenta are summed FIRST, then the mass is taken from the sum)
    lep_E = ak.sum(E, axis=1)
    lep_px = ak.sum(lep_pt * np.cos(phi), axis=1)
    lep_py = ak.sum(lep_pt * np.sin(phi), axis=1)
    lep_pz = ak.sum(lep_pt / np.tan(2.0 * np.arctan(np.exp(-eta))), axis=1)  # pT * sinh(eta), written via eta

    # invariant mass from E^2 = p^2 + m^2; guard against small negative values from resolution effects
    Mass_2 = lep_E**2 - (lep_px**2 + lep_py**2 + lep_pz**2)
    Mass = ak.where(Mass_2 > 0, np.sqrt(Mass_2), 0)
    return Mass
```

<a id='processing'></a>

## Reading and selecting events

One function handles both streams, branching on the sample name it was given.

### The jet stream

Requires a trigger to have fired, and for the single-jet path additionally demands a jet above 500 GeV. ATLAS records only a tiny fraction of collisions, so without a trigger requirement we would be measuring the trigger's preferences rather than the detector's behaviour. The 500 GeV threshold
puts us safely above the turn-on region where trigger efficiency is changing rapidly.

### The lepton stream

Flavour is identified through `lep_type`, which stores the PDG code of each lepton 11 for an electron, 13 for a muon. Summing over the event gives 22 for an electron pair and 26 for a muon pair, which is how the flavour selection is done here. The matching dielectron or dimuon trigger is then
required.

Charge works the same way: summing `lep_charge` gives 0 for an opposite-sign pair and $\pm 2$ for a same-sign one.

Both leptons must then exceed 25 GeV comfortably above the trigger thresholds, and high enough that fake leptons from jets are strongly suppressed. Finally the invariant mass is computed and attached to
the array as a new field, so the plotting functions can treat it exactly like any other branch.

```python
def process_data_and_mc(samples, loop, what):
  ALL = []
  #Loop over all ROOT files in the sample
  for file in tqdm(samples[what]["list"], desc=f"Processing {what}"):
    tree = uproot.open(file + ":analysis")
    numevents = tree.num_entries
    #Iterate over the tree in chunks
    for data in tree.iterate(library="ak",entry_start = int( round(numevents * fraction * loop ,0) ),entry_stop = int( round(numevents * fraction *(loop+1),0) )):
      # ---- jet stream: require a trigger, and stay above its turn-on region ----
      if (what == 'qcdjets_IF') | (what == 'qcdjets_IG') | (what == 'qcdjets_GRL'):
        if tit == 'J':
          data = data[data.trigJ]
          data = data[ak.any(data.jet_pt > 500, axis=1)]   # well above the single-jet turn-on
        else:
          data = data[data.trigMET]
          data = data[(data.met > 300)]
      # ---- lepton stream: flavour, charge, then kinematics ----
      elif (what == '2lep_IF') | (what == '2lep_IG') | (what == '2lep_GRL'):
        # lep_type holds the PDG code: 11 = electron, 13 = muon.
        # Summed over the event: 22 = electron pair, 26 = muon pair.
        st = ak.sum(data.lep_type, axis = 1)
        if til == 'lep':
          data = data[(st == 22) | (st == 26)]
          data = data[ (data.trigDE) | (data.trigDM) ]
        elif til == 'e':
          data = data[st == 22]
          data = data[data.trigDE]
        else:
          data = data[st == 26]
          data = data[data.trigDM]
        # Summed charge: 0 for an opposite-sign pair, +-2 for a same-sign pair
        sc = ak.sum(data.lep_charge, axis = 1)
        if tis == 'op':
          data = data[sc == 0] # opp
        else:
          data = data[(sc == 2) | (sc == -2)] # same
        data = data[ak.sum(data.lep_pt > 25, axis=1) >= 2]   # above threshold, suppresses fakes
        data['Mass'] = mass(data.lep_e,data.lep_pt,data.lep_eta,data.lep_phi)

      ALL.append(data)   # every chunk stays in memory - keep 'fraction' modest
  return ALL
```

## Running everything

`run()` is the driver: it processes whichever streams `ty` selects, then fires off the full battery of comparison plots.

The variables are declared `global` so that `p3` can reach them by name it looks up `lepdata_GRL`, `lepdata_IG` and so on directly rather than receiving them as arguments.

### Why the same variable is plotted three times

Look at the MET calls: the same quantity is plotted over 0-5000 GeV, then 0-1000, then 50-400. This is not redundancy, it is the method.

Missing transverse energy is the classic pathology detector. A real event has MET of order tens of GeV; a *broken* event one with a noise burst in the calorimeter, or written out incompletely can have MET in the thousands. So the wide range is a pathology hunt, deliberately absurd in scale, looking for
the handful of events that should not exist. The narrow ranges then zoom into the physics region to check that the three selections agree where it actually matters.

The same logic applies to the mass plots: full range, then to 200 GeV, then tight around the Z at 50-150 GeV, and to jet multiplicity at 0-50 and 0-10.


```python
def run():
  # declared global so p3() can look them up by name rather than receiving them as arguments
  global lepdata_IF, lepdata_IG, lepdata_GRL
  global qcddata_IF, qcddata_IG, qcddata_GRL
  global fraction

  if (ty == 'lep') | (ty == 'both'):
    #Data analysis
    fraction = 0.001
    lepdata_IF = process_data_and_mc(samples, 0, '2lep_IF')
    lepdata_IG = process_data_and_mc(samples, 0, '2lep_IG')
    fraction = 0.001     # only this value takes effect - the calls above are commented out
    lepdata_GRL = process_data_and_mc(samples, 0, '2lep_GRL')

    p3('Mass', 'Inv_Mass_all', 'Inv_Mass [GeV]', 0, 600, 200, '2lep inveriant mass (Norm)', 1)
    p3('Mass', 'Inv_Mass_to_200', 'Inv_Mass [GeV]', 0, 200, 100, '2lep inveriant mass (Norm)', 1)


  if (ty == 'jet') | (ty == 'both'):
    #Data analysis
    fraction = 0.001
    qcddata_IF = process_data_and_mc(samples, 0, 'qcdjets_IF')
    qcddata_IG = process_data_and_mc(samples, 0, 'qcdjets_IG')
    fraction = 0.001
    qcddata_GRL = process_data_and_mc(samples, 0, 'qcdjets_GRL')

    p3('jet_n', 'Number of jets', 0, 50, 90, 'QCD number of jets (Norm)', 0)
    p3('jet_n', 'Number of jets', 0, 10, 30, 'QCD number of jets (Norm)', 0)
    p3('met', 'met [GeV]', 0, 5000, 100, 'QCD MET (Norm)', 0)
    p3('met', 'met [GeV]', 0, 1000, 100, 'QCD MET (Norm)', 0)
    p3('met', 'met [GeV]', 50, 400, 100, 'QCD MET (Norm)', 0)

  return 0
```
